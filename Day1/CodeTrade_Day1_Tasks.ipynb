***4.Tip Calculator***

```python
def calculate_tip(bill, tip_percent):
    tip = (bill*tip_percent)/100
    total = bill + tip
    return tip, total

tip1, total1 = calculate_tip(1000,10)
tip2, total2 = calculate_tip(800,15)
tip3 , total3 = calculate_tip(545,5)
print(f"Bill: 1000| Tip:{tip1} | Total:{total1}")
print(f"Bill: 1000| Tip:{tip2} | Total:{total2}") 
print(f"Bill: 1000| Tip:{tip3} | Total:{total3}")
```

***5.Word Frequency***

```python
sen = input("enter the sentence here")
words = sen.lower().split()
frequency = {}
for word in words: 
    if word in frequency: 
        frequency[word]+=1
    else: 
        frequency[word]=1
print("Word Frequency:\n")
for word, count in frequency.items():
    print(f"{word}:{count}")
```

***6.Simple Calculator***

```python
def add(a, b):
    return a + b
def subtract(a, b):
    return a - b
def multiply(a, b):
    return a * b
def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"

    return a / b
operations = {
    "add": add,
    "subtract": subtract,
    "multiply": multiply,
    "divide": divide
}
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

choice = input("Choose operation (add/subtract/multiply/divide): ").lower()
if choice in operations:
    result = operations[choice](num1, num2)

    print("\nResult:", result)

else:
    print("Invalid operation")
```

***7.Grade Classifier***

```python
students = { "Anuj":92, "Anshula":95, "Rohan":65, "Anshu": 85}
for name , score in students.items(): 
    if score >=90:
        grade="A"
    elif score >=75: 
        grade="B"
    elif score >= 60: 
        grade="C"
    else: 
        grade = "D"
    print(name,score,grade)
```

***9.Mini Contact Book***

```python
contacts = [
    {
        "name": "Anuj",
        "phone": "9876543210",
        "email": "anuj@gmail.com"
    },

    {
        "name": "Rahul",
        "phone": "9123456780",
        "email": "rahul@gmail.com"
    },

    {
        "name": "Priya",
        "phone": "9988776655",
        "email": "priya@gmail.com"
    }
]
def find_contact(name):
    for contact in contacts:
        if contact["name"].lower() == name.lower():
            return contact
    return "Contact not found"
search_name = input("Enter contact name: ")
result = find_contact(search_name)
print("\nResult:",result)
```
