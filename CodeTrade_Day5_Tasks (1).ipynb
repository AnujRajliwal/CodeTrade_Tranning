***Task 1***

```python
class Student: 
    school_name = "ABC"
    def __init__(self,name,rollno,marks):
        self.name = name
        self.rollno = rollno
        self.marks = marks
    def average(self):
        return sum(self.marks) / len(self.marks)
    def grade(self):
        avg = self.average()
        if avg >= 90:
            return "A"
        elif avg >= 75:
            return "B"
        elif avg >= 60:
            return "C"
        elif avg >= 40:
            return "D"
        else:
            return "F"
    def __str__(self):
        return (
            f"School: {Student.school_name}\n"
            f"Name: {self.name}\n"
            f"Roll No: {self.rollno}\n"
            f"Marks: {self.marks}\n"
            f"Average: {self.average():.2f}\n"
            f"Grade: {self.grade()}\n"
        )
student1 = Student(
    "Anuj",
    101,
    [90, 85, 88]
)
student2 = Student(
    "Rahul",
    102,
    [70, 75, 80]
)
student3 = Student(
    "Priya",
    103,
    [50, 65, 60]
)
print(student1)
print(student2)
print(student3)
```

***Task 3***

```python
import csv
students = [
    ["Anuj", 90, 85, 88],
    ["Rahul", 70, 75, 80],
    ["Priya", 60, 65, 70]
]
with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "math", "science", "english"])
    writer.writerows(students)
print("students.csv created successfully")
with open("students.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)
    results = []
    for row in reader:
        name = row[0]
        math = int(row[1])
        science = int(row[2])
        english = int(row[3])
        average = (math + science + english) / 3
        if average >= 90:
            grade = "A"
        elif average >= 75:
            grade = "B"
        elif average >= 60:
            grade = "C"
        else:
            grade = "D"
        results.append([name, round(average, 2), grade])
with open("results.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "average", "grade"])
    writer.writerows(results)
print("results.csv created successfully")
```

```python
from typing import Optional
def add(a: float, b: float) -> float:
    """Returns addition of two numbers"""
    return a + b
def subtract(a: float, b: float) -> float:
    """Returns subtraction of two numbers"""
    return a - b
def multiply(a: float, b: float) -> float:
    """Returns multiplication of two numbers"""
    return a * b
def divide(a: float, b: float) -> Optional[float]:
    """Returns division result"""
    if b == 0:
        return None
    return a / b
def power(base: float, exp: float) -> float:
    """Returns power value"""
    return base ** exp
def modulo(a: int, b: int) -> Optional[int]:
    """Returns remainder"""
    if b == 0:
        return None
    return a % b
print(add(10, 5))
print(subtract(10, 5))
print(multiply(10, 5))
print(divide(10, 2))
print(power(2, 3))
print(modulo(10, 3))
```

```python
class LibraryItem:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
    def describe(self):
        return (
            f"{self.title} by "
            f"{self.author} ({self.year})"
        )
class Book(LibraryItem):
    def __init__(self, title, author, year, pages):
        super().__init__(title, author, year)
        self.pages = pages
    def describe(self):
        return (
            f"{self.title} by "
            f"{self.author} "
            f"({self.year}) "
            f"- {self.pages} pages"
        )
class EBook(LibraryItem):
    def __init__(self, title, author, year, file_size_mb):
        super().__init__(title, author, year)
        self.file_size_mb = file_size_mb
    def describe(self):
        return (
            f"{self.title} by "
            f"{self.author} "
            f"({self.year}) "
            f"- {self.file_size_mb}MB"
        )
book1 = Book(
    "Python Basics",
    "Anuj",
    2024,
    300
)
book2 = Book(
    "AI Guide",
    "Rahul",
    2023,
    250
)
ebook1 = EBook(
    "ML Notes",
    "Priya",
    2025,
    15
)
ebook2 = EBook(
    "Data Science",
    "Amit",
    2022,
    20
)
library_items = [
    book1,
    book2,
    ebook1,
    ebook2
]
for item in library_items:
    print(item.describe())
print(
    isinstance(book1, LibraryItem)
)
```

***Task 6***

```python
import json
def save_config(data, filename):
    with open(filename, "w") as file:
        json.dump(data, file, indent=4)
    print("Config saved successfully")
def load_config(filename):
    with open(filename, "r") as file:
        data = json.load(file)
    return data
def update_config(filename, key, value):
    data = load_config(filename)
    data[key] = value
    save_config(data, filename)
    print("Config updated successfully")
config = {
    "model": "Linear Regression",
    "learning_rate": 0.01,
    "epochs": 10
}
save_config(config, "config.json")
update_config("config.json", "epochs", 20)
updated_data = load_config("config.json")
print(updated_data)
```

```python
import pandas as pd
data = {
    "name": [
        "Anuj", "Rahul", "Priya", "Amit", "Neha",
        "Riya", "Karan", "Simran", "Vikas", "Pooja"
    ],
    "city": [
        "Delhi", "Mumbai", "Delhi", "Pune", "Mumbai",
        "Delhi", "Pune", "Mumbai", "Delhi", "Pune"
    ],
    "math_score":
        [90, 70, 85, 60, 95, 88, 72, 65, 80, 75],

    "science_score":
        [85, 75, 80, 65, 90, 82, 70, 60, 78, 72],

    "english_score":
        [88, 68, 84, 62, 91, 86, 74, 63, 79, 76]
}
df = pd.DataFrame(data)
print("\nAverage Scores:")
print(df[[
    "math_score",
    "science_score",
    "english_score"
]].mean())
df["total"] = (
    df["math_score"] +
    df["science_score"] +
    df["english_score"]
)
print("\nTop Student:")
print(df.loc[df["total"].idxmax()])
print("\nStudents from each city:")
print(df["city"].value_counts())
print("\nMath score above 75:")
print(df[df["math_score"] > 75])
print("\nTop 3 Students:")
print(df.nlargest(3, "total"))
```

```python
import math
class Fraction:
    def __init__(self, numerator, denominator):
        self.numerator = numerator
        self.denominator = denominator
    def __str__(self):
        return f"{self.numerator}/{self.denominator}"
    def __add__(self, other):
        num = (
            self.numerator * other.denominator +
            other.numerator * self.denominator
        )
        den = (
            self.denominator * other.denominator
        )
        gcd = math.gcd(num, den)

        return Fraction(
            num // gcd,
            den // gcd
        )
    def __eq__(self, other):

        return (
            self.numerator * other.denominator ==
            other.numerator * self.denominator
        )
    def __lt__(self, other):

        return (
            self.numerator * other.denominator <
            other.numerator * self.denominator
        )
f1 = Fraction(1, 2)
f2 = Fraction(1, 3)
f3 = Fraction(2, 4)
f4 = Fraction(1, 2)
print("Addition:", f1 + f2)
print("Equal:", f3 == f4)
print("Less Than:", f1 < f2)
```

```python
import csv
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity
class Inventory:
    def __init__(self):
        self.products = []
    def add_product(self, product):
        self.products.append(product)
    def total_value(self):
        total = 0
        for product in self.products:
            total += (
                product.price *
                product.quantity
            )
        return total
    def find_product(self, name):
        for product in self.products:
            if product.name.lower() == name.lower():
                return product
        return None
    def save_to_csv(self, filename):
        with open(filename, "w", newline="") as file:
            writer = csv.writer(file)
            writer.writerow([
                "name",
                "price",
                "quantity"
            ])
            for product in self.products:
                writer.writerow([
                    product.name,
                    product.price,
                    product.quantity
                ])
    def load_from_csv(self, filename):
        with open(filename, "r") as file:
            reader = csv.reader(file)
            next(reader)
            for row in reader:
                product = Product(
                    row[0],
                    float(row[1]),
                    int(row[2])
                )
                self.add_product(product)
inventory = Inventory()
inventory.add_product(
    Product("Laptop", 50000, 2)
)
inventory.add_product(
    Product("Phone", 25000, 3)
)
inventory.save_to_csv("inventory.csv")
print(
    "Total Value:",
    inventory.total_value()
)
```
