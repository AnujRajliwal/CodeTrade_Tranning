***Task 1***

```python
student = {
    "name": "Anuj",
    "course": "AIML",
    "skills": ["Python", "Pandas", "NumPy"],
    "city": "Mahendragarh"
}
def create_profile(profile: dict) -> str:

    skills = ", ".join(profile["skills"])

    return (
        f"Name: {profile['name']}\n"
        f"Course: {profile['course']}\n"
        f"Skills: {skills}\n"
        f"City: {profile['city']}"
    )
print(create_profile(student))
```

```python
class Learner:
    def __init__(self, name, course):

        self.name = name
        self.course = course
    def show_details(self):

        return (
            f"{self.name} is learning "
            f"{self.course}"
        )
student1 = Learner(
    "Anuj",
    "Python"
)
print(student1.show_details())
```

```python
import pandas as pd
data = {

    "name": [
        "Anuj","Rahul","Priya","Amit","Neha"
    ],

    "city": [
        "Delhi","Mumbai","Delhi","Pune","Mumbai"
    ],

    "score": [
        92,70,85,60,95
    ]
}
df = pd.DataFrame(data)
selected_columns = df[
    ["name", "score"]
]
print("\nSelected Columns:")
print(selected_columns)
filtered_data = df[
    (df["score"] > 80)
    &
    (df["city"] == "Delhi")
]
print("\nFiltered Rows:")
print(filtered_data)
```

```python
import numpy as np
data = {

    "name": [
        "Anuj",
        "Rahul",
        "Priya",
        "Amit"
    ],

    "score": [
        92,
        np.nan,
        85,
        60
    ],

    "city": [
        "Delhi",
        "Mumbai",
        np.nan,
        "Pune"
    ]
}
df = pd.DataFrame(data)
print("\nMissing Values:")
print(df.isnull().sum())
drop_data = df.dropna()

print("\nAfter dropna():")
print(drop_data)
fill_data = df.fillna({

    "score": 0,
    "city": "Unknown"
})

print("\nAfter fillna():")
print(fill_data)
```

```python
array = np.array([10, 20, 30, 40, 50])
mask = array > 25
filtered_values = array[mask]
print("Original Array:")
print(array)
print("\nBoolean Mask:")
print(mask)
print("\nFiltered Values:")
print(filtered_values)
scaled_array = array + 5
print("\nBroadcasted Array:")
print(scaled_array)
def cosine_similarity(vec1, vec2):
    dot_product = np.dot(vec1, vec2)
    norm_vec1 = np.linalg.norm(vec1)
    norm_vec2 = np.linalg.norm(vec2)
    similarity = (
        dot_product /
        (norm_vec1 * norm_vec2)
    )
    return similarity
v1 = np.array([1, 2, 3])
v2 = np.array([1, 2, 3])
v3 = np.array([1, 0, 0])
v4 = np.array([0, 1, 0])
print("\nCosine Similarity 1:")
print(cosine_similarity(v1, v2))

print("\nCosine Similarity 2:")
print(cosine_similarity(v3, v4))
```
